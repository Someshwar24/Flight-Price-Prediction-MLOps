## ***5. Feature Engineering & Data Pre-processing***

```python
#Separate features and target
gender_X = gender_df[["company", "age"]]
gender_y = gender_df["gender"]

print("Input features:", gender_X.columns.tolist())
print("Target variable:", gender_y.name)
```

Uses company and age as input features and gender as the target. The code column is excluded because it is an identifier, and name is excluded from this baseline because nearly every name is unique.

```python
#Encode company into numerical features
from sklearn.preprocessing import OneHotEncoder

gender_encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
gender_company_encoded = gender_encoder.fit_transform(gender_X[["company"]])

gender_company_df = pd.DataFrame(gender_company_encoded, columns=gender_encoder.get_feature_names_out(["company"]), index=gender_X.index)
gender_company_df["age"] = gender_X["age"]

gender_company_df.head()
```

```python
#Split the data
from sklearn.model_selection import train_test_split

gender_X_train, gender_X_test, gender_y_train, gender_y_test = train_test_split(
    gender_company_df, gender_y, test_size=0.2, random_state=42, stratify=gender_y
)

print("Training records:", len(gender_X_train))
print("Testing records:", len(gender_X_test))
```

```python
#Scale age
from sklearn.preprocessing import StandardScaler

gender_scaler = StandardScaler()
gender_X_train = gender_X_train.copy()
gender_X_test = gender_X_test.copy()

gender_X_train[["age"]] = gender_scaler.fit_transform(gender_X_train[["age"]])
gender_X_test[["age"]] = gender_scaler.transform(gender_X_test[["age"]])

gender_X_train.head()
```

## ***6. ML Model Implementation***

```python
# Train classification models
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

gender_X = gender_df[["company", "age"]]
gender_y = gender_df["gender"]

gender_X_train, gender_X_test, gender_y_train, gender_y_test = train_test_split(gender_X, gender_y, test_size=0.2, random_state=42, stratify=gender_y)

gender_preprocessor = ColumnTransformer([("company", OneHotEncoder(handle_unknown="ignore"), ["company"]), ("age", make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), ["age"])])

gender_models = {"Logistic Regression": LogisticRegression(max_iter=1000), "Decision Tree": DecisionTreeClassifier(random_state=42), "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42)}
gender_results = []

for name, model in gender_models.items():
    gender_pipeline = make_pipeline(gender_preprocessor, model)
    gender_pipeline.fit(gender_X_train, gender_y_train)
    gender_predictions = gender_pipeline.predict(gender_X_test)
    gender_results.append({"Model": name, "Accuracy": accuracy_score(gender_y_test, gender_predictions)})
    print(f"\n{name}\n", classification_report(gender_y_test, gender_predictions, zero_division=0))

gender_results_df = pd.DataFrame(gender_results).sort_values("Accuracy", ascending=False)
gender_results_df
```

```python
# 5.2: Compare model accuracy
sns.barplot(data=gender_results_df, x="Model", y="Accuracy", hue="Model", legend=False)
plt.title("Classification Model Accuracy Comparison")
plt.ylabel("Test Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=15)
plt.show()
```

The three classes have similar support, so class imbalance is not the obvious cause of poor performance.

Logistic Regression performs slightly better than the other models, but its accuracy is still low.

company and age may not contain enough information to predict the recorded gender labels reliably.

The model's results are only slightly better than the roughly 33% accuracy expected from random guessing across three balanced classes.

```python
# Feature importance experiment: include name as a categorical feature
gender_X_name = df[["company", "age", "name"]]
gender_y_name = df["gender"]

gender_X_train_name, gender_X_test_name, gender_y_train_name, gender_y_test_name = train_test_split(
    gender_X_name, gender_y_name, test_size=0.2, random_state=42, stratify=gender_y_name
)

gender_name_preprocessor = ColumnTransformer([
    ("categorical", OneHotEncoder(handle_unknown="ignore"), ["company", "name"]),
    ("age", StandardScaler(), ["age"])
])

gender_name_model = make_pipeline(gender_name_preprocessor, LogisticRegression(max_iter=1000))
gender_name_model.fit(gender_X_train_name, gender_y_train_name)

gender_name_predictions = gender_name_model.predict(gender_X_test_name)

print(classification_report(gender_y_test_name, gender_name_predictions, zero_division=0))
```

#### 2. Cross- Validation & Hyperparameter Tuning

```python
#Tune Logistic Regression
from sklearn.model_selection import GridSearchCV

gender_logistic_pipeline = make_pipeline(
    gender_preprocessor,
    LogisticRegression(max_iter=1000)
)

gender_grid = GridSearchCV(
    gender_logistic_pipeline,
    {"logisticregression__C": [0.01, 0.1, 1, 10]},
    cv=5,
    scoring="f1_macro"
)

gender_grid.fit(gender_X_train, gender_y_train)

print("Best parameters:", gender_grid.best_params_)
print("Best cross-validation macro F1:", gender_grid.best_score_)
```

```python
# 5.4: Evaluate the tuned model
gender_best_model = gender_grid.best_estimator_
gender_tuned_predictions = gender_best_model.predict(gender_X_test)

print("Test accuracy:", accuracy_score(gender_y_test, gender_tuned_predictions))
print(classification_report(
    gender_y_test, gender_tuned_predictions, zero_division=0
))
```
